```python
from sklearn.metrics import ( # metrics - инструменты для оценки точности модели  
    accuracy_score,  # Доля правильных прогнозов от общего числа прогнозов.
    precision_score,  # Точность: Способность классификатора не помечать как положительные отрицательные образцы.
    recall_score,  # Полнота: Способность классификатора находить все положительные образцы.
    f1_score,  # F1-мера: Гармоническое среднее между точностью и полнотой.
    roc_auc_score,  # Площадь под ROC-кривой (AUC): Мера способности классификатора различать классы.
    confusion_matrix,  # Матрица ошибок: Таблица, показывающая правильные и неправильные прогнозы.
    classification_report,  # Отчет о классификации: Содержит точность, полноту, F1-меру и поддержку для каждого класса.
    log_loss,  # Логарифмическая потеря (cross-entropy loss): Функция потерь, используемая в задачах классификации.
    average_precision_score, # Average precision (AP): Сводная метрика, отражающая соотношение точности и полноты для разных порогов классификации. Для задач с несбалансированными классами.
    brier_score_loss, # Brier score: Измерение точности вероятностных прогнозов.
    matthews_corrcoef,  # Коэффициент корреляции Мэтьюза: Мера качества бинарной (или многоклассовой) классификации, особенно при несбалансированных классах.
    cohen_kappa_score, # Kappa Cohen: Мера согласия между двумя оценивающими (классификаторами), корректирует случайное соглашение.
    hamming_loss, # Hamming loss: Доля неверно классифицированных меток (для multi-label классификации).
    jaccard_score, # Jaccard index: Мера сходства между наборами прогнозируемых и фактических меток (для multi-label классификации).
    zero_one_loss # Zero-one loss: Просто подсчитывает количество неверно классифицированных образцов.
)
```

```python
from sklearn.metrics import (
    mean_absolute_error,  # Средняя абсолютная ошибка (MAE): Средняя величина абсолютных разностей между прогнозами и фактическими значениями.  Легко интерпретируется.
    mean_squared_error,  # Среднеквадратическая ошибка (MSE): Среднее значение квадратов разностей между прогнозами и фактическими значениями. Чувствительна к выбросам.
    mean_squared_log_error, # Среднеквадратичная логарифмическая ошибка (MSLE): Используется, когда целевая переменная имеет экспоненциальный рост (например, продажи, арендная плата).
    median_absolute_error,  # Медианная абсолютная ошибка: медианное значение всех абсолютных разностей между прогнозами и фактическими значениями.  Менее чувствительна к выбросам, чем MAE.
    r2_score,  # Коэффициент детерминации (R^2): Мера того, насколько хорошо модель соответствует данным.  Принимает значения от 0 до 1, где 1 означает идеальное соответствие.
    explained_variance_score,  # Оценка объясненной дисперсии: Измеряет долю дисперсии в целевой переменной, которая объясняется моделью.
    max_error, # Максимальная остаточная ошибка: Показывает самое большое различие между прогнозируемым и фактическим значением.
    mean_poisson_deviance, # Среднее отклонение Пуассона: Полезно для регрессии с отсчетами Пуассона.

    #The following scale dependent regression error metrics are rescaled versions of the mean absolute error to improve their interpretability across various scales.
    mean_absolute_percentage_error, # Средняя абсолютная процентная ошибка (MAPE).
    d2_absolute_error_score # d2_absolute_error_score
    d2_pinball_score # d2_pinball_score
)
```
