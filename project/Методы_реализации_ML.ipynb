## Кореляции 

3 - основных вида корреляции
- Линейная (измеряется коэффициентом Пирсона).
- Монотонная (измеряется коэффициентами Спирмена или Кендалла).
- Непараметрическая (для порядковых данных или данных с выбросами, используются коэффициенты Спирмена, Кендалла).


### Используйте метод .corr() для DataFrame или Series:
- Пирсон (линейная): df.corr(method='pearson') или просто df.corr().
- Спирмен (монотонная): df.corr(method='spearman').
- Кендалл (монотонная, для малых выборок): df.corr(method='kendall').
### Пример: 
df['столбец1'].corr(df['столбец2'], method='spearman') рассчитает корреляцию Спирмена между двумя столбцами.
### формулы
корреляция Пирсона применяется именно к числовым данным. Если точнее, то к количественным (непрерывным) переменным, таким как рост, вес, температура, цена.
- Пирсона = r_xy = Σ(x_i - Mx)(y_i - My) / [(n-1) σx σy]
Вычисление Пирсона для рангов. Сначала все значения превращаются в их порядковые номера (ранги), а затем считается обычная корреляция Пирсона для этих рангов.
- Cпирмена = 1 - [6 × Σ(d_i)²] / [n(n² - 1)] # где d_i = rank(x_i) - rank(y_i) — разность рангов
Сравнение всех пар. Берутся все пары точек, и считается, в скольких парах порядок по X и Y совпадает, а в скольких — противоположен.
- Кендалл = (n_c - n_d) / √[(n_0 - n_1)(n_0 - n_2)] # где n_c — согласованные пары, n_d — несогласованные

```python
a = df['age'].corr(df['hours.per.week'], method='pearson') # 	Корреляция Пирсона, применённая к числам 
b = df['age'].corr(df['hours.per.week'], method='spearman') # Вычисление Пирсона для рангов. 
# Сначала все значения превращаются в их порядковые номера (ранги), а затем считается обычная корреляция Пирсона для этих рангов.
c = df['age'].corr(df['hours.per.week'], method='kendall')
# Разница результатов 
print(*[round(x, 2) for x in [a, b, c]]) 
```
